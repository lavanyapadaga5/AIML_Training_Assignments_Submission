**Numerical missing value imputation**

Imputation means replacing missing values with estimated or selected values

Types of Imputation:
* Mean Imputation=sum of values/Number of values
* Median Imputation=Median means middle value after getting sorted
* Mode Imputation=The value which Occurs the most

```python
import pandas as pd 
import numpy as np
from sklearn.impute import SimpleImputer,KNNImputer

##CReate a sample dataset
data={
    'Age':[25,30,np.nan,4550,np.nan,35],
    'Salary':[50000,60000,5500,np.nan,'London','Paris','New York',np.nan],
    'City':['New York','London',np.nan,'London','Paris','New York',np.nan]
}
df=pd.dataframe(data)
print("Original data:\n",df,"\n")
#make a copy to preserve original data 
df_imputed=df.copy()
#NUmerical Imputation
Num_cols=['Age','Salary']
num_imputer=SimpleImputer(strategy='median')
df_imputed[num_cols]=num_imputer.fit_transform(df_imputed[num_cols])
# CAtegorical Imputation
cat_cols=['City']
cat_imputer=SimpleImputer(strategy='most_frequent')
df_imputed[cat_cols]=cat_imputer.fit_transform(df_imputed[cat_cols])
print("Imputed data",df_imputed)
#save the clean dataset
df_imputed.to_csv('Dataset saved successfully')
```

**3. Scaling – Standardization and Normalization**

Some machine learning algorithms are sensitive to these differences in scale.

Scaling transforms numerical values into a comparable range.

There are two important techniques.

**A. Min-Max Normalization**

t transforms values into a range, usually 0 to 1.

Formula:
\[
X_{scaled}=\frac{X-X_{min}}{X_{max}-X_{min}}
\]

```python
from sklearn.preprocessing import MinMaxScaler

data = {
    "Age": [20, 21, 22],
    "Salary": [20000, 30000, 40000]
}

df = pd.DataFrame(data)

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print(scaled_df)
```

B)Standardization(StandardScaler)

Standard transforms data based on its mean and standard deviation

\[
Z=\frac{X-\mu}{\sigma}
\]

Where:
- \(X\) = Original value.
- \(\mu\) = Mean.
- \(\sigma\) = Standard deviation.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

standardized_data = scaler.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print(standardized_df)
```

After standardization:
- Mean is approximately 0.
- Standard deviation is approximately 1.
The values can be negative or positive.

**Difference between normalization and standardization**

**Normalization**

* Usually range 0 to 1	
* Uses minimum and maximum	
* Sensitive to outliers	
* Uses MinMaxScaler

**Standardization**

* Mean approximately 0
* Uses mean and standard deviation
* Also affected by outliers
* Uses StandardScaler

**When do you need scaling?**

caling is useful for algorithms such as:
- K-Nearest Neighbors (KNN).
- Support Vector Machines (SVM).
- Logistic Regression.
- K-Means clustering.
- Neural networks.

Decision trees and Random Forests generally don't require feature scaling.

Four employees earn between ₹20,000 and ₹25,000.

But one employee earns ₹5,00,000.

That value is very different from the others.

This unusual value is a potential outlier.

An outlier is an observation that is unusually far from the majority of observations.

**Why do outliers matter?**

20000, 22000, 25000, 23000, 500000

```python
import numpy as np

salary = [20000, 22000, 25000, 23000, 500000]

print(np.mean(salary))
```

But most employees earn around ₹20,000–₹25,000.

The ₹5,00,000 salary has increased the average significantly.

This is why outlier analysis is important.

However, an outlier is not necessarily an error. It could be a genuine high salary, an important transaction, or a rare but valid observation.

**B. Detect outliers using IQR**

IQR stands for Interquartile Range.

The formula:
\[
IQR = Q3-Q1
\]

```python
import pandas as pd

df = pd.DataFrame({
    "Salary": [
        20000, 22000, 25000, 23000,
        24000, 26000, 21000, 500000
    ]
})
```

**Calculating Q1 and Q3**

```python
Q1 = df["Salary"].quantile(0.25)

Q3 = df["Salary"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

**Calculate IQR**

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

**Calculate boundaries**

```python
lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)

print("Upper Limit:", upper_limit)
```

**Identify outliers**

```python
outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

print(outliers)
```

**What should you do after finding an outlier?**

There are three common approaches.

**Remove the outlier**

```python
df_cleaned = df[
    (df["Salary"] >= lower_limit) &
    (df["Salary"] <= upper_limit)
]
```

**Cap the outlier**

```python
df["Salary"] = df["Salary"].clip(
    lower=lower_limit,
    upper=upper_limit
)
```

**Keep the outlier**

If ₹5,00,000 is a genuine salary, removing it could destroy useful information.

The correct decision depends on the business problem and the source of the data.
