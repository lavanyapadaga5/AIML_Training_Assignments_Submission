
# Practical Exercise: Student Grade Calculator

## Project Practice

### Objective
Create a Python program to calculate a student's grade based on their marks.

### Requirements

1. Store a student's name in a variable.
2. Store the student's marks in a variable.
3. Check whether the student has passed or failed.
4. Assign a grade based on the student's marks.
5. Display the student's name, marks, result, and grade.

### Grading Criteria

| Marks | Grade |
|---|---|
| 90 - 100 | A |
| 75 - 89 | B |
| 60 - 74 | C |
| 40 - 59 | D |
| Below 40 | Fail |

### Expected Output

```text
Student Name: Lavanya
Marks: 92
Result: Passed
Grade: A
```

### Additional Challenge

- What happens if a student enters marks greater than 100?
- What happens if a student enters negative marks?
- Validate the marks before assigning a grade.

### Learning Outcomes

- Understand Python variables.
- Practice conditional statements.
- Use comparison operators.
- Implement decision-making logic.
- Display formatted output.


````python

# Step 1: Get Student Details

student_name = input("Enter Student Name: ")
student_marks = int(input("Enter Student Marks: "))


# Step 2: Validate Marks and Calculate Grade

if student_marks < 0 or student_marks > 100:
    grade = "Invalid Marks"
    result = "Invalid"

elif student_marks >= 90:
    grade = "A"
    result = "Passed"

elif student_marks >= 75:
    grade = "B"
    result = "Passed"

elif student_marks >= 60:
    grade = "C"
    result = "Passed"

elif student_marks >= 40:
    grade = "D"
    result = "Passed"

else:
    grade = "Fail"
    result = "Failed"


# Step 3: Display Student Details

print("\n----- Student Details -----")

print("Student Name:", student_name)
print("Student Marks:", student_marks)
print("Result:", result)
print("Grade:", grade)
````


# Practical Exercise: Student Marks Analyzer

## Objective
Create a Python program to process multiple student records using loops.

## Requirements

1. Create a list containing student names.
2. Create another list containing their marks.
3. Use a for loop to iterate through students.
4. Display each student's name and marks.
5. Calculate the total marks.
6. Calculate the average marks.
7. Count how many students passed and failed.

## Sample Dataset

| Student | Marks |
|---|---:|
| Lavanya | 92 |
| Rahul | 75 |
| Priya | 35 |
| Arun | 65 |
| Sneha | 28 |

## Expected Analysis

- Total Students: 5
- Total Marks: 295
- Average Marks: 59
- Passed Students: 3
- Failed Students: 2

## Learning Outcomes

- Understand for loops.
- Practice lists.
- Use conditional statements inside loops.
- Calculate totals and averages.


````python

students = ["Lavanya", "Rahul", "Priya", "Arun", "Sneha"]

marks = [92, 75, 35, 65, 28]

total_marks = 0
passed = 0
failed = 0

for i in range(len(students)):

    print("Student:", students[i])
    print("Marks:", marks[i])

    total_marks += marks[i]

    if marks[i] >= 40:
        passed += 1
        print("Result: Passed")
    else:
        failed += 1
        print("Result: Failed")

average_marks = total_marks / len(students)

print("\n----- SUMMARY -----")
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Passed Students:", passed)
print("Failed Students:", failed)
highest_marks = marks[0]
topper = students[0]

for i in range(len(students)):
    if marks[i] > highest_marks:
        highest_marks = marks[i]
        topper = students[i]

print("topper is:",topper)

lowest_marks = marks[0]

for i in range(len(students)):
    if marks[i] <lowest_marks:
        lowest_marks = marks[i]
        person = students[i]

print("Person with lowest marks:",person)

print("\nStudents Above Average:")

for i in range(len(students)):

    if marks[i] > average_marks:
        print(students[i], "-", marks[i])


````

**Variable Scope in Python**

Python mainly has four types of scope:
1. Local scope
2. Global scope
3. Enclosing scope
4. Built-in scope

**Local Variables**

A variable created inside a function is called a local variable.

````python
def calculate_total():
    total = 1000
    print(total)

calculate_total()
````

**Global Variables**

A variable created outside a function is called a global variable.

````python
company = "Vitelglobal"

def display_company():
    print(company)

display_company()
print(company)
````

**Global Keyword**

What if you want to modify a global variable inside a function?

````python
count = 10

def update_count():
    global count
    count = 20

update_count()

print(count)
````

**The global keyword tells Python that you want to modify the global variable instead of creating a local one.

Important: Avoid using global variables excessively. They make programs harder to maintain and test.**

**Enclosing Scope**

This happens when you have a function inside another function.

````python
def outer_function():

    message = "Hello"

    def inner_function():
        print(message)

    inner_function()

outer_function()
````
